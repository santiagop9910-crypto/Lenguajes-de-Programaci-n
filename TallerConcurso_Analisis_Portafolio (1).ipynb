## 2. ¿Qué es un Análisis de Portafolio de Inversión? 📚


✍️ El análisis de portafolio de inversión es el proceso de estudiar cómo se comportan los diferentes activos financieros que una persona tiene o planea tener, con el fin de encontrar el mejor equilibrio entre riesgo y rendimiento.
La idea principal es que no basta con mirar un activo por separado, sino analizar cómo se relaciona con los demás dentro del portafolio, ya que una buena combinación puede reducir el riesgo total sin sacrificar tanto la ganancia esperada.

En este análisis se calculan indicadores como el rendimiento esperado, que muestra cuánto se puede ganar en promedio, y el riesgo, que mide qué tanto pueden variar esos resultados. También se consideran aspectos como la correlación entre activos y la eficiencia del portafolio, que busca encontrar las combinaciones más rentables para un nivel de riesgo dado.

Hoy en día, herramientas como Python y Google Colab facilitan este tipo de estudios gracias a librerías como yfinance, pandas y matplotlib, que permiten descargar datos reales del mercado, calcular métricas automáticamente y graficar los resultados para una mejor interpretación.


## 3. Investigación: Tipos de Análisis de Portafolio 🔍

✍️ Para el desarrollo del código que se presentará en el concurso, se seleccionaron tres tipos de análisis que permiten estudiar portafolios con uno o dos activos, priorizando la facilidad de programación y la claridad al mostrar los resultados. Estos son: análisis de rentabilidad y riesgo, diversificación y correlación, y comparación de portafolio de 1 activo vs. 2 activos.


**Análisis de rentabilidad y riesgo:**
Este análisis es el punto de partida, ya que permite medir cuánto se gana y cuánto se arriesga al invertir en un activo.
El objetivo es encontrar un equilibrio entre el rendimiento esperado y el riesgo asumido. Para ello, se calculan medidas como la media (rendimiento promedio), la desviación estándar (variación del rendimiento) y el ratio de Sharpe, que compara la ganancia obtenida frente al riesgo tomado.

En Python y Google Colab, se descargan los precios históricos de los activos con yfinance, se calculan los rendimientos diarios con pct_change(), y luego se usan mean() y std() para obtener las medidas de tendencia y dispersión. Finalmente, se pueden graficar los resultados para visualizar la relación entre riesgo y retorno.

**Diversificación y correlación:** Este tipo de análisis se centra en entender cómo se mueven los activos entre sí y cómo esa relación puede ayudar a reducir el riesgo total del portafolio.
Si dos activos se mueven en direcciones opuestas o tienen una correlación baja, combinarlos puede ayudar a estabilizar los resultados.

El objetivo es aprovechar la diversificación, es decir, no depender de un solo activo para obtener buenos resultados. En Python, este análisis se hace fácilmente con el método corr() de pandas, y los resultados se pueden visualizar con mapas de calor (heatmaps) que muestran de forma gráfica qué tan relacionados están los activos.



**Comparación de portafolio de 1 activo vs. 2 activos**  
Este análisis es ideal para mostrar los beneficios de la diversificación de manera práctica.
Consiste en comparar el desempeño de un activo individual con el de un portafolio que combina dos activos en distintas proporciones. A medida que se cambian los pesos de inversión, se observa cómo varía el rendimiento esperado y la volatilidad del portafolio.

En Python, esto se logra fácilmente definiendo un rango de pesos y calculando para cada uno el rendimiento y la volatilidad total. El resultado se puede graficar como una curva de riesgo–rendimiento, donde se ve claramente que la combinación de activos puede generar menor riesgo con un rendimiento similar o mayor.
