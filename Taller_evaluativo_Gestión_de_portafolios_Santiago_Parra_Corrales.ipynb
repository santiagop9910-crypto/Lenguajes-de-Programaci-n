La empresa ha recopilado los siguientes datos sobre cuatro clientes

```python
# Listas para almacenar información de clientes
clientes = ["Juan Perez","Maria Rodriguez","Carlos Lopez","Ana Gomez"]
montos_invertidos = [10000, 15000, 20000, 18000]  # en dólares
rendimientos = [5.5, 6.0, 4.0, 7.5]  # en porcentaje

```

## **PARTE 2 Uso de diccionarios y condicionales**

1. **Diccionario Portafolio**
A continuación se encontrará un diccionario llamado portafolio que almacena la información de cada cliente con su monto invertido y rendimiento; llamando cada clave por el nombre del cliente.

```python
portafolio = {
 "Juan Perez": {"monto_invertido": 10000, "rendimiento": 5.5}, "Maria Rodriguez": {"monto_invertido": 15000, "rendimiento": 6.0},"Carlos Lopez": {"monto_invertido": 20000, "rendimiento": 4.0},"Ana Gomez": {"monto_invertido": 18000, "rendimiento": 7.5}
}
```


2. **Evaluación del rendimiento de los clientes**
La empresa quiere identificar qué clientes han tenido un rendimiento superior al 5%

```python
for cliente, datos in portafolio.items():
    if datos["rendimiento"] > 5:
        print(f"{cliente} tiene un rendimiento de {datos["rendimiento"]}%")
```

## **PARTE 3 Calculos con ciclos**

1. **Calcular el Rendimiento Total del Portafolio**
La empresa desea calcular el rendimiento total del portafolio, que se define como la suma de todos los rendimientos en dólares.


```python
rendimiento_total = 0
for cliente, datos in portafolio.items():
    rendimiento_totalcliente = datos["monto_invertido"] * (datos["rendimiento"] / 100)
    print(f"{cliente} tiene un rendimiento total de {rendimiento_totalcliente} dólares")
    rendimiento_total+= rendimiento_totalcliente
print(f"\nEl rendimiento total del portafolio es de {rendimiento_total} dólares")
```

2. **Cliente con Mayor Rendimiento**
La empresa necesita saber qué cliente ha obtenido el mayor rendimiento en términos absolutos.

```python
cliente_max=""
rendimiento_max= 0
for cliente, datos in portafolio.items():
    rendimiento_totalcliente = datos["monto_invertido"] * (datos["rendimiento"] / 100)
if rendimiento_totalcliente > rendimiento_max:
    rendimiento_max = rendimiento_totalcliente
    cliente_max = cliente
    print(f"el cliente con mayor rendimiento es {cliente} con {rendimiento_totalcliente} dólares")
```
